# بررسی محیط اجرای پروژه
این نوت‌بوک را از پوشهٔ اصلی پروژه اجرا کنید.

```python
import sys, platform, importlib.metadata as md
print("Python:", sys.version)
print("Platform:", platform.platform())
for name in ["numpy", "pandas", "matplotlib", "scikit-learn", "opencv-python", "scikit-image", "jupyter"]:
    try: print(f"{name}: {md.version(name)}")
    except md.PackageNotFoundError: print(f"{name}: نصب نیست")
```
